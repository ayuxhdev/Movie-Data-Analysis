### Categorizing the `Vote_Average` Column

We have to cut the `Vote_Average` values and make 4 categories : `popular` `average` `below_average` `not_popular` to describe it more using 'categorize_col()' function provided above . 

```python
def categorize_col(df, col, labels ) :
    
    edges = [df[col].describe()['min'] , 
    df[col].describe()['25%'] , 
    df[col].describe()['50%'] ,
    df[col].describe()['75%'] , 
    df[col].describe()['max']]
    
    df[col + '_Category'] = pd.cut(
        df[col],
        edges,
        labels = labels,
        duplicates = 'drop',
        include_lowest = True
    )
    
    return df
```

```python
labels = ["not popular" , "below average" , "average" , "popular"]

categorize_col(df , 'Vote_Average' , labels)

df['Vote_Average_Category'].unique()
```

### Inserting the Data

```python
import mysql.connector
from getpass import getpass

conn = mysql.connector.connect(
    host = "localhost",
    port = 3306,
    user = "root",
    password = getpass("Enter your MySQL Password:"),
    database = "movie_analysis"
)

cursor = conn.cursor()

try :
    # Prevent accidental duplicate imports
    for table in ["movies","genres","movie_genres"] :
        cursor.execute(f"SELECT COUNT(*) from {table}")
        count = cursor.fetchone()[0]
        
        if count > 0 :
            raise ValueError(
                f"{table} is not empty. Import Stopped."
            )
            
    genre_ids = {}
    relationships = []
    
    for _, row in df.iterrows() :
        # Insert the movies 
        cursor.execute(
            """INSERT INTO movies(
                release_year,
                title,
                popularity,
                vote_count,
                vote_average,
                original_language,
                vote_average_category
            )
            VALUES(%s, %s, %s, %s, %s, %s, %s)""",
            (
                int(row["Release_Date"]),
                row["Title"],
                float(row["Popularity"]),
                int(row["Vote_Count"]),
                float(row["Vote_Average"]),
                row["Original_Language"],
                str(row["Vote_Average_Category"])
            )
        )
        
        movie_id = cursor.lastrowid
        
        # Split genres and remove extra spaces
        movie_genres = {
            genre.strip()
            for genre in str(row["Genre"]).split(",")
            if genre.strip()
        }
        
        for genre in movie_genres :
            # Create each genre only once 
            if genre not in genre_ids :
                cursor.execute(
                    "INSERT INTO genres (genre_name) VALUES(%s)",
                    (genre,)
                )
                
                genre_ids[genre] = cursor.lastrowid
                
            relationships.append((movie_id, genre_ids[genre]))
    
    # Insert all movie-genre relationships
    cursor.executemany(
        """
        INSERT INTO movie_genres(movie_id,genre_id)
        VALUES(%s,%s)
        """,
        relationships
    )
    
    conn.commit()
    
    print("IMPORT SUCCESSFUL!")
    print("Movies Inserted:", len(df))
    print("Unique Genres:", len(genre_ids))
    print("Genre Relationships:", len(relationships))
        
except Exception:
    conn.rollback()
    raise

finally:
    cursor.close()
    conn.close()
```
